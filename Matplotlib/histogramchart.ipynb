```python
# Import library
import matplotlib.pyplot as plt
import numpy as np
```

```python

data = np.random.normal(loc=80, scale=10, size=100)  # 200 data acak
data = np.clip(data,0,100)

plt.hist(data, bins=10, color="orange", edgecolor="black")
plt.title("Distribusi Nilai Ujian")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show()
```
