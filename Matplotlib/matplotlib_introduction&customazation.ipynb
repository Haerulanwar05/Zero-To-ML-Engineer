```python
import matplotlib.pyplot as plt   # mengimport sebuah library matplotlib
import numpy as np
```

```python
x = np.array(['roti cokelat','roti keju','roti sarikaya','roti moka'])
y1 = np.array([17,12,30,3])
y2 = np.array([4,16,7,13])
y3 = np.array([12,12,3,1])

# mmbuat sebuah data dumb dengan numpy
```

```python
# menyimpan settingan seperti template
line_stye = dict(marker=".",
              markersize=30,
              markerfacecolor="#5e74e0",
              markeredgecolor="#0c0f1f",
              linewidth=3,
              linestyle='solid')


# menampilkan visualisasi dengan matplotlib dengan sub bawaan pyplot
plt.plot(x,y1,color="#0c0f1f", **line_stye)       # fungsi untuk membuat sebuah visualisasi dan penempatan data 
plt.plot(x,y2,color="#508f8a", **line_stye)
plt.plot(x,y3,color="#2bc249", **line_stye)

plt.show()

# untuk markersize bisa disingkat dengan ms (opsional)
# untuk markerfacecolor bisa disingkat dengan mfc (opsional)
# kita bisa mengatur settingan dari linestye yang telah kita buat dengan menggunakan fungsi dict() dan untuk membuka nya kita membutuhkan 2 bintang
```
