```python
# Import library
import matplotlib.pyplot as plt
import numpy as np
```

```python

tinggi1 = np.array([150, 160, 165, 170, 175, 180])
berat1 = np.array([50, 55, 60, 65, 70, 78])

tinggi2 = np.array([140, 155, 160, 177, 179, 183])
berat2 = np.array([45, 50, 56, 64, 74, 85])

plt.scatter(tinggi1, berat1, color="darkgreen",
                            s = 100,
                            alpha = 0.5,
                            label="Sample A")

plt.scatter(tinggi2, berat2, color="darkred",
                            s = 100,
                            alpha = 0.5,
                            label="Sample B")

plt.xlabel("Tinggi (cm)")
plt.ylabel("Berat (kg)")
plt.title("Hubungan Tinggi dan Berat Badan")
plt.legend()
plt.show()
```
