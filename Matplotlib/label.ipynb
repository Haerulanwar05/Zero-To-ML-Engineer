```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([13, 15, 20, 30])

plt.title("Class size", fontsize=20,
                        fontweight="bold",
                        color="#eb4d62",
                        family="Arial")

plt.xlabel("Year", fontsize=15,
                        fontweight="bold",
                        color="#eb4d62",
                        family="Arial")

plt.ylabel("Students", fontsize=15,
                        fontweight="bold",
                        color="#eb4d62",
                        family="Arial")


plt.tick_params(axis="both",
                colors="#eb4d62")

plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)

plt.xticks(x)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.linspace(0, 10, 100)   # 100 titik dari 0 sampai 10
b = np.sin(a)

plt.plot(a, b)
plt.show()
```

```python
import matplotlib.pyplot as plt

c = [1, 2, 3, 4, 5]
produk_a = [10, 20, 25, 30, 40]
produk_b = [5, 15, 20, 22, 28]

plt.plot(c, produk_a, label="Produk A")
plt.plot(c, produk_b, label="Produk B")
plt.xticks(c)

plt.xlabel("Minggu")
plt.ylabel("Unit Terjual")
plt.title("Perbandingan Penjualan Produk")
plt.legend()
plt.show()
```
