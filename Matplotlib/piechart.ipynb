```python
# Import library
import matplotlib.pyplot as plt
import numpy as np
```

```python

label = ["Makanan", "Transportasi", "Hiburan", "Tabungan"]
persentase = np.array([35, 20, 15, 30])

colors = ["skyblue", "lightgray", "yellow", "green"]

plt.pie(persentase, labels=label, 
                    autopct="%1.1f%%",
                    colors= colors,
                    explode=[0, 0, 0, 0.1])
plt.title("Alokasi Pengeluaran Bulanan")
plt.show()
```
