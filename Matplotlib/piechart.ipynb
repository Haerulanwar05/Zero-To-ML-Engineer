```python
# import library 
import matplotlib.pyplot as plt
import numpy as np

```

```python
label = ["Makanan", "Transportasi", "Hiburan", "Tabungan"]
persentase = [35, 20, 15, 30]
colors = ["lightgray","skyblue","darkgreen","red"]
plt.pie(persentase, labels=label,
                    autopct="%1.1f%%",
                    colors=colors,
                    explode=[0.1,0,0,0],
                    startangle=20,
                    shadow=True)
plt.title("Alokasi Pengeluaran Bulanan")
plt.show()
```
