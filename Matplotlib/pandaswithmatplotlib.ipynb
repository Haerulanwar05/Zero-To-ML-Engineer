```python
# import library
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
type_counts = df['Type1'].value_counts(ascending=True)

plt.barh(type_counts.index, type_counts.values,
                                        color="lightgray",
                                        edgecolor="black")

plt.xlabel("count")
plt.ylabel("Type")
plt.title("# of Count pokemon type primary")
plt.tight_layout()
plt.show()
```
