```python
# import library

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statistics as sts 
```

```python
pd.set_option('display.max_rows', None)

mart = pd.read_excel('supermarket_sales.xlsx')
mart = mart[['Gender','Payment','Unit price','Quantity','Total','gross income']]
mart.head(10)
```

```python
sns.kdeplot(data=mart, x='Total')
```

```python
# membat sebuah KDE untuk semua variabel dengan data numerik di dataframe (mart)
plt.figure(figsize=(15,5))
sns.kdeplot(data=mart)
```

```python
sns.kdeplot(data=mart,x='Total',bw_adjust=0.2) # mengatur kehalusan KDE
```

```python
sns.kdeplot(data=mart, x='Total', hue='Payment')
```

```python
sns.kdeplot(data=mart, x='Total', hue='Payment', multiple='stack') 
```

```python
sns.kdeplot(data=mart, x='Total', log_scale=True)
```

```python


# ❌ Skala Linear Standar (Data menumpuk di sisi kiri, ekor panjang tak terbaca)
plt.figure(figsize=(10, 4))
sns.kdeplot(data=mart, x='Total')
plt.title("Tanpa log_scale (Skala Linear Standar)")
plt.show()

# ✅ Menggunakan log_scale (Distribusi terlihat proporsional dan jelas)
plt.figure(figsize=(10, 4))
sns.kdeplot(data=mart, x='Total', log_scale=True, color='teal')
plt.title("Dengan log_scale=True (Skala Logaritmik Basis 10)")
plt.show()
```

```python
sns.kdeplot(data=mart, x='Total', hue='Payment', multiple='stack', linewidth=5,edgecolor='white',palette='CMRmap',alpha=0.5) 
```

```python
sns.kdeplot(data=mart, x='Unit price', y='gross income',hue='Gender', fill=True, levels=6, thresh=0.5)
```

```python
plt.colormaps()
```

```python
plt.figure(figsize=(8, 6))
sns.kdeplot(
    data=mart,
    x='Unit price',
    y='gross income',
    fill=True,              # Mengisi warna pada gradasi cincin kontur
    cmap='viridis',         # Peta warna gradasi densitas kontinu
    thresh=0.05,            # Memotong area densitas di bawah 5%
    cbar=True               # Menampilkan colorbar skala nilai densitas
)
plt.title('2D Bivariate KDE Plot - Hubungan Unit Price vs Gross Income')
plt.show()
```
