```python
# import library

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statistics as sts 
```

```python
iris = sns.load_dataset('iris')
iris.head()
```

```python
sns.jointplot(data=iris, x='sepal_length', y='petal_length'\
            ,kind='resid')
```

```python
sns.jointplot(data=iris, x='sepal_length', y='petal_length', hue='species')
```

```python
sns.jointplot(data=iris, x='sepal_length', y='petal_length',color='red',hue='species',palette='Set2')
```

```python
sns.jointplot(data=iris, x='sepal_length', y='petal_length'
                # ,color='red'
                # ,hue='species'
                # ,palette='Set2'
                ,joint_kws=dict(color='black',marker='+')
                ,marginal_kws=dict(color='purple',kde=True))
```

```python
sns.jointplot(data=iris, x='sepal_length', y='petal_length'
              ,height=5
              ,ratio=3
            #   ,space=5
              ,marginal_ticks=True)
```

```python
l = sns.jointplot(data=iris, x='sepal_length', y='petal_length',joint_kws=dict(color='red'))
l.plot_joint(sns.kdeplot)
l.plot_joint(sns.rugplot, height=0.1,color='red')
```
