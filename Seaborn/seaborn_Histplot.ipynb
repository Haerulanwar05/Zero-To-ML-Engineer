```python
# import library

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statistics as sts 
```

```python
pd.set_option('display.max_rows', None)

mart = pd.read_excel('supermarket_sales.xlsx')
mart = mart[['Gender','Payment','Unit price','Quantity','Total','gross income']]
mart.head(10)
```

```python
# basic chart Histogram

sns.histplot(data=mart,y='Total')
```

```python
# basic chart Histogram

sns.histplot(data=mart,x='Total',binwidth=100)
```

```python
sns.histplot(data=mart,x='Total',bins=np.arange(0,1100,50))
```

```python
# menggati jumlah Bins dan intervall nya
plt.figure(figsize=(15,5))
sns.histplot(data=mart,x='Total',bins=np.arange(0,1100,50))
plt.xticks(np.arange(0,1100,50))
```

```python
# mengkombnasi dengan KDE
sns.histplot(data=mart,x='Total', kde=True)
```

```python
sns.histplot(data=mart,x='Total',hue='Payment',multiple='stack') # mengatur bentuk sebuah histogram denga element
```

```python
sns.histplot(data=mart,x='Total',hue='Payment',element='poly') # mengatur bentuk sebuah histogram denga element
```

```python
sns.histplot(data=mart,x='Total',stat='count',shrink=0.9) # membuat sebuah gap agar bins tidak menyatu dengan shrink
```

```python
mart.head()
```

```python
sns.histplot(data=mart,x='Total',y='gross income') # membuat sebuah histogram dengan 2 variable
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(
    data=mart,
    x='Unit price',
    y='gross income',
    bins=30,
    cbar=True,              # Menampilkan colorbar skala frekuensi kepadatan
    cmap='viridis'
)
plt.show()
```
