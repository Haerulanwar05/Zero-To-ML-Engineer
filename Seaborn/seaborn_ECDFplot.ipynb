```python
# import library

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statistics as sts 
```

```python
pd.set_option('display.max_rows', None)

mart = pd.read_excel('supermarket_sales.xlsx')
mart = mart[['Gender','Payment','Unit price','Quantity','Total','gross income']]
mart.head(10)
```

```python
# membuat sebuah ECDF Plot
sns.ecdfplot(data=mart,x='gross income',hue='Gender')
```

```python
# membuat sebuah ECDF Plot
sns.ecdfplot(data=mart,y='gross income',hue='Gender')
```

```python
plt.figure(figsize=(10, 5))
# Membuat ECDF untuk kolom Total belanja
sns.ecdfplot(data=mart, x='Total')
plt.title('ECDF Plot - Distribusi Kumulatif Total Belanja')
plt.xlabel('Total Belanja ($)')
plt.ylabel('Proporsi Kumulatif (0 - 1.0)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.ecdfplot(
    data=mart,
    x='Total',
    complementary=True,     # Menampilkan proporsi nilai yang LEBIH BESAR dari X
    color='crimson',
    linewidth=2
)
plt.title('Complementary ECDF (CCDF) - Proporsi Transaksi di Atas Nilai X')
plt.xlabel('Total Belanja ($)')
plt.ylabel('Proporsi Customer Berbelanja > X')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
