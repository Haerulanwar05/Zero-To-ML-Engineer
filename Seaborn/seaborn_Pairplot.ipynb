```python
# import library

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statistics as sts 
```

```python
iris = sns.load_dataset('iris')
iris.head()
```

```python
sns.pairplot(data=iris)
```

```python
sns.pairplot(data=iris,diag_kind='kde')
```

```python
sns.pairplot(data=iris,kind='reg')
```

```python
sns.pairplot(data=iris,hue='species')
```

```python
sns.pairplot(data=iris,x_vars=['sepal_width','petal_width','petal_length'],y_vars= ['sepal_length','sepal_width'],diag_kind=None)
```

```python
sns.pairplot(data=iris,corner=True)
```

```python
sns.pairplot(data=iris,diag_kws=dict(color='green',kde=True),plot_kws=dict(color='black',marker=10, s=60))
```

```python
l = sns.pairplot(data=iris,plot_kws=dict(color='green'))
l.map_upper(sns.kdeplot)
l.map_lower(sns.kdeplot)
```

```python
# Memaksa diagonal menggunakan histogram meskipun menggunakan hue
sns.pairplot(
    data=iris,
    hue='species',
    diag_kind='hist',
    palette='tab10'
)
plt.show()
```

```python
# Menambahkan garis regresi linier pada setiap panel bivariate
sns.pairplot(
    data=iris,
    hue='species',
    kind='reg',             # Menambahkan garis regresi linier per kategori
    palette='Dark2'
)
plt.show()
```

```python
sns.pairplot(
    data=iris,
    hue='species',
    corner=True,            # Hanya menampilkan segitiga bawah kanvas
    palette='Set1'
)
plt.show()
```

```python
# Hanya menganalisis sepal_length dan sepal_width
sns.pairplot(
    data=iris,
    vars=['sepal_length', 'sepal_width'],
    hue='species',
    palette='husl'
)
plt.show()
```

```python
sns.pairplot(
    data=iris,
    hue='species',
    diag_kind='kde',
    corner=True,
    palette='Set2',
    plot_kws=dict(alpha=0.7, s=45, edgecolor='black', linewidth=0.5),
    diag_kws=dict(fill=True, alpha=0.35, linewidth=2)
)
plt.show()
```
