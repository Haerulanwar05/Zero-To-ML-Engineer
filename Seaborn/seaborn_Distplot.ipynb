```python
# import library

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statistics as sts 
```

```python
pd.set_option('display.max_rows', None)

mart = pd.read_excel('supermarket_sales.xlsx')
mart = mart[['Gender','Payment','Unit price','Quantity','Total','gross income','Branch']]
mart.head(10)
```

```python
sns.displot(data=mart, y='Total', rug=True, rug_kws={'height':0.05}, kde=True,hue='Payment',multiple='stack',col='Branch',row='Gender')
```

```python
sns.displot(data=mart, x='Total',kind='ecdf', hue='Payment',col='Branch',row='Gender')
```
