```python
# import library

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import dataset
mart = pd.read_excel('supermarket_sales.xlsx')
mart.head()
```

```python
sns.swarmplot(data=mart, y='Total')
```

```python
sns.swarmplot(data=mart, y='Total', x='Payment', hue='Gender',dodge=True)
```

```python

plt.figure(figsize=(15,5))
sns.swarmplot(data=mart, y='Total', x='Payment', marker='*',size=8, hue='Payment')
```
