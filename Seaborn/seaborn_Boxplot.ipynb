```python
# import library

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import dataset
mart = pd.read_excel('supermarket_sales.xlsx')
mart.head()
```

```python
# membuat sebuah box plot

sns.boxplot(x='Payment', y='Total', hue='Gender', data=mart, palette='viridis')
```

```python
print(plt.colormaps())
```

```python

plt.grid(True, axis='y',linewidth=0.5, linestyle='solid' )
sns.boxplot(y='Total', data=mart, width=0.3)

```

```python
sns.boxplot(x='Payment', y='Total',hue='Payment', data=mart)  # sebuah box plot dengan kategori variabel
```

```python
sns.boxplot(x='Payment', y='Total',hue='Gender', data=mart)  # sebuah box plot dengan kategori variabel
```

```python
sns.boxplot(x='Payment', y='Total',hue='Gender', data=mart,showmeans=True,meanprops={"marker":"o", "markerfacecolor":"white", "markeredgecolor":"black"}
)  # sebuah box plot dengan kategori variabel
```

```python
sns.boxplot(x='Total', y='Payment',hue='Gender', data=mart,showmeans=True,meanprops={"marker":"o", 
                                                                                     "markerfacecolor":"white", 
                                                                                     "markeredgecolor":"black",
                                                                                     "markersize" : "5"},
palette='turbo',
linewidth= 2
)  # sebuah box plot dengan kategori variabel
```

```python
# membuat sebuah boxplot universal

sns.boxplot(data=mart)
```
