```python
# import library

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# import dataset
mart = pd.read_excel('supermarket_sales.xlsx')
mart.head()
```

```python
sns.stripplot(data=mart, y='Total')
```

```python
sns.stripplot(data=mart, y='Total',jitter=0.2)  # agar tidak overlap perlu menambahkan sebuah jitter
```

```python
sns.stripplot(data=mart, y='Total',linewidth=1)
```

```python
sns.stripplot(data=mart, y='Total', x='Payment', hue='Gender',jitter=0.2,linewidth=0.5)
```

```python
sns.stripplot(data=mart, y='Total', x='Payment', hue='Gender',jitter=0.2,linewidth=0.5,dodge=True)
```

```python
sns.stripplot(data=mart,y='Total', x='Payment',jitter=0.1, color='black')
sns.boxplot(data=mart,y='Total', x='Payment', color='grey')
```

```python
sns.stripplot(data=mart,y='Total', x='Payment',jitter=0.1, color='black')
sns.violinplot(data=mart,y='Total', x='Payment', color='grey')
```
